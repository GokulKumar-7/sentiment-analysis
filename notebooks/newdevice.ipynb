```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
data = pd.read_csv('/content/newdevice.csv')

# Plotting
fig, ax1 = plt.subplots(figsize=(10,6))

# Bar for Active Users
bars = ax1.bar(
    data['Device category'],
    data['Active users'],
    color='#FFA500',  # orange
    label='Active Users'
)
ax1.set_ylabel('Active Users', color='#FFA500')
ax1.tick_params(axis='y', labelcolor='#FFA500')

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    ax1.annotate(f'{int(height):,}', xy=(bar.get_x() + bar.get_width()/2, height),
                 xytext=(0, 5), textcoords='offset points', ha='center', fontsize=9, fontweight='bold')

# Line 1: Avg Engagement Time per Active User
ax2 = ax1.twinx()
line1 = ax2.plot(
    data['Device category'],
    data['Average engagement time per active user'],
    color='#8B0000', marker='o', linewidth=2, label='Avg Engagement Time per Active User (s)'
)

# Add annotations to Line 1
for i, value in enumerate(data['Average engagement time per active user']):
    ax2.annotate(f'{round(value,1)}s', xy=(i, value),
                 xytext=(0, 5), textcoords="offset points", ha='center', fontsize=9, color='#8B0000')

# Line 2: Engaged Sessions per Active User
line2 = ax2.plot(
    data['Device category'],
    data['Engaged sessions per active user'],
    color='#006400', marker='s', linewidth=2, label='Engaged Sessions per Active User'
)

# Add annotations to Line 2
for i, value in enumerate(data['Engaged sessions per active user']):
    ax2.annotate(f'{round(value,1)}', xy=(i, value),
                 xytext=(0, -15), textcoords="offset points", ha='center', fontsize=9, color='#006400')

ax2.set_ylabel('Seconds / Sessions per User', color='black')
ax2.tick_params(axis='y', labelcolor='black')

# Title and layout
plt.title('Device Category: Active Users, Engagement Time & Sessions per User', fontsize=12, fontweight='bold')
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
plt.legend(lines + lines2, labels + labels2, loc='upper right', frameon=False)

plt.tight_layout()
plt.show()
```
