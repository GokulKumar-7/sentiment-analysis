Analyze the page paths where Sessions and Views are high but Engaged Sessions per Active User is low

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 🔹 Load the dataset (Update with actual file path)
file_path = "/content/All segment final.csv"  # Replace with your actual file path
df = pd.read_csv(file_path)

# 🔹 Filter pages where Sessions and Views are high but Engagement is low
high_traffic_low_engagement = df[(df["Sessions"] > df["Sessions"].mean()) &
                                 (df["Views"] > df["Views"].mean()) &
                                 (df["Engaged sessions per active user"] < df["Engaged sessions per active user"].mean())]

# 🔹 Sort by Sessions & Views (descending) and select **top 10 pages**
high_traffic_low_engagement = high_traffic_low_engagement.sort_values(by=["Sessions", "Views"], ascending=False).head(10)

# 🔹 Visualization: Dual-Axis Bar Chart (Sessions, Views vs. Engagement)
fig, ax1 = plt.subplots(figsize=(14, 7))

# 🔹 Plot Sessions and Views (Bar Chart)
bar_width = 0.4  # Width of bars
x_labels = high_traffic_low_engagement["Page path and screen class"]
x_indexes = range(len(x_labels))

bars1 = ax1.bar(x_indexes, high_traffic_low_engagement["Sessions"], width=bar_width, label="Sessions", alpha=0.7, color='orange')
bars2 = ax1.bar([x + bar_width for x in x_indexes], high_traffic_low_engagement["Views"], width=bar_width, label="Views", alpha=0.7, color='green')

ax1.set_ylabel("Sessions / Views")
ax1.set_xticks([x + bar_width/2 for x in x_indexes])
ax1.set_xticklabels(x_labels, rotation=45, ha="right")

# 🔹 Plot Engaged Sessions per Active User (Line Chart - Secondary Axis)
ax2 = ax1.twinx()
line, = ax2.plot([x + bar_width/2 for x in x_indexes],
                 high_traffic_low_engagement["Engaged sessions per active user"],
                 label="Engaged Sessions per Active User",
                 marker='o', linestyle='dashed', color='red')

ax2.set_ylabel("Engaged Sessions per Active User")

# 🔹 Add values on top of bars
for bar in bars1:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(), round(bar.get_height(), 1), ha='center', va='bottom', fontsize=10)

for bar in bars2:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(), round(bar.get_height(), 1), ha='center', va='bottom', fontsize=10)

# 🔹 Add values on top of the line chart (3 decimal places)
for i, txt in enumerate(high_traffic_low_engagement["Engaged sessions per active user"]):
    ax2.text(x_indexes[i] + bar_width/2, txt, f"{txt:.3f}", ha='center', va='bottom', fontsize=10, color='red')

# 🔹 Titles & Labels
ax1.set_title("High Traffic Pages with Low Engagement", fontsize=14)
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.show()

# 🔹 Print the filtered dataset
print("\n🔹 Pages with High Traffic but Low Engagement:")
high_traffic_low_engagement
```

Top 10 Pages with High Entrances and High Bounce Rate -

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 🔹 Load the dataset (Update with actual file path)
file_path = "/content/All segment final.csv"  # Replace with your actual file path
df = pd.read_csv(file_path)

# 🔹 Filter pages with High Entrances and High Bounce Rate
high_entrance_high_bounce = df[(df["Entrances"] > df["Entrances"].mean()) &
                               (df["Bounce rate"] > df["Bounce rate"].mean())]

# 🔹 Sort by Entrances (descending) and select **top 10 pages**
high_entrance_high_bounce = high_entrance_high_bounce.sort_values(by=["Entrances"], ascending=False).head(10)

# 🔹 Visualization: Dual-Axis Bar Chart (Entrances, Sessions vs. Bounce Rate)
fig, ax1 = plt.subplots(figsize=(14, 7))

# 🔹 Plot Entrances and Sessions (Bar Chart)
bar_width = 0.4  # Width of bars
x_labels = high_entrance_high_bounce["Page path and screen class"]
x_indexes = range(len(x_labels))

bars1 = ax1.bar(x_indexes, high_entrance_high_bounce["Entrances"], width=bar_width, label="Entrances", alpha=0.7, color='blue')
bars2 = ax1.bar([x + bar_width for x in x_indexes], high_entrance_high_bounce["Sessions"], width=bar_width, label="Sessions", alpha=0.7, color='green')

ax1.set_ylabel("Entrances / Sessions")
ax1.set_xticks([x + bar_width/2 for x in x_indexes])
ax1.set_xticklabels(x_labels, rotation=45, ha="right")

# 🔹 Plot Bounce Rate (Line Chart - Secondary Axis)
ax2 = ax1.twinx()
line, = ax2.plot([x + bar_width/2 for x in x_indexes],
                 high_entrance_high_bounce["Bounce rate"],
                 label="Bounce Rate",
                 marker='o', linestyle='dashed', color='red')

ax2.set_ylabel("Bounce Rate (%)")

# 🔹 Add values on top of bars
for bar in bars1:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(), round(bar.get_height(), 1), ha='center', va='bottom', fontsize=10)

for bar in bars2:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(), round(bar.get_height(), 1), ha='center', va='bottom', fontsize=10)

# 🔹 Add values on top of the line chart (2 decimal places for Bounce Rate)
for i, txt in enumerate(high_entrance_high_bounce["Bounce rate"]):
    ax2.text(x_indexes[i] + bar_width/2, txt, f"{txt:.2f}", ha='center', va='bottom', fontsize=10, color='red')

# 🔹 Titles & Labels
ax1.set_title("High Entrance Pages with High Bounce Rate", fontsize=14)
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.show()

# 🔹 Print the filtered dataset
print("\n🔹 Pages with High Entrances and High Bounce Rate:")
high_entrance_high_bounce
```

Analyze the Exits vs. Views per Session for top pages

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df is your DataFrame with the necessary columns
# Remove "/yeokens/sign-out" and "/yeokens/sign-in" from the dataset
df_filtered = df[~df["Page path and screen class"].isin(["/yeokens/sign-out", "/yeokens/sign-in"])]

# Calculate total exits and average views per session by page
page_summary = df_filtered.groupby("Page path and screen class")[["Exits", "Views per session"]].sum().reset_index()

# Sort the data by Exits in descending order to find the top pages with most exits
page_summary_sorted = page_summary.sort_values(by="Exits", ascending=False)

# Select top 10 pages with the highest exits (after removing "/yeokens/sign-out" and "/yeokens/sign-in")
top_pages = page_summary_sorted.head(10)

# Create figure and axis
fig, ax1 = plt.subplots(figsize=(12, 7))

# Bar Chart for Exits
bars = sns.barplot(x="Page path and screen class", y="Exits", data=top_pages, color="blue", ax=ax1)
ax1.set_xlabel("Page Path and Screen Class")
ax1.set_ylabel("Exits")
ax1.set_title("Exits vs Views per Session")

# Add labels on bars with 3 decimal points for Exits
for bar in bars.patches:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
             f'{bar.get_height():.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold', color='black')

# Create secondary y-axis for Views per Session (Line Chart)
ax2 = ax1.twinx()
ax2.plot(top_pages["Page path and screen class"], top_pages["Views per session"],
         marker="o", linestyle="-", color="red", linewidth=2, label="Views per Session")

ax2.set_ylabel("Views per Session", color="red")

# Add labels for Views per Session on the line
for i, value in enumerate(top_pages["Views per session"]):
    ax2.text(i, value, f'{value:.3f}', color='red', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Rotate x-axis labels vertically to prevent overlap
plt.xticks(rotation=45, ha="center")

# Show plot
plt.tight_layout()
plt.show()

# Display the table with the top 10 pages
print("\nTop 10 Pages with High Exits and Views per Session (Excluding '/yeokens/sign-out' and '/yeokens/sign-in'):")
print(top_pages)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your GA data
df = pd.read_csv("/content/All segment final.csv")

# Remove "/yeokens/sign-in" from the dataset
df_filtered = df[df["Page path and screen class"] != "/yeokens/sign-in"]

# Get the top 10 pages based on highest number of sessions
top_sessions = df_filtered.sort_values(by="Sessions", ascending=False).head(10)

# Display table
print("\nTop 10 Pages - User Journey & Session Depth Analysis (Excluding '/yeokens/sign-in')\n")
print(top_sessions[["Page path and screen class", "Sessions", "Engaged sessions per active user"]])

# Set up the figure
fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar plot for Sessions (orange)
barplot = sns.barplot(x="Page path and screen class", y="Sessions", data=top_sessions,
                      color="orange", ax=ax1)

# Add session values on top of bars
for bar in barplot.patches:
    height = bar.get_height()
    ax1.annotate(f"{int(height)}",
                 (bar.get_x() + bar.get_width() / 2, height + 2000),
                 ha='center', va='bottom', fontsize=10, color='black', fontweight='bold')

# Line plot for Engaged sessions per active user (red)
ax2 = ax1.twinx()
sns.lineplot(x="Page path and screen class", y="Engaged sessions per active user", data=top_sessions,
             color="red", marker="o", ax=ax2)

# Annotate engaged sessions values (with special positioning)
overlap_pages = ["/yeokens/dairy-go-round", "/yeokens/get-banking", "/yeokens/hub", "/yeokens/welcome"]

for i, (page, val) in enumerate(zip(top_sessions["Page path and screen class"], top_sessions["Engaged sessions per active user"])):
    offset = -12 if page in overlap_pages else 8
    va = 'top' if offset < 0 else 'bottom'
    ax2.annotate(f"{val:.2f}", (i, val), textcoords="offset points", xytext=(0, offset),
                 ha='center', va=va, fontsize=10, color='red', fontweight='bold')

# Axis labels and title
ax1.set_xlabel("Page Path", fontsize=12)
ax1.set_ylabel("Sessions", fontsize=12, color="orange")
ax2.set_ylabel("Engaged sessions per active user", fontsize=12, color="red")
ax1.set_xticklabels(top_sessions["Page path and screen class"], rotation=45, ha="right", fontsize=10)

# Remove legends
ax1.get_legend().remove() if ax1.get_legend() else None
ax2.get_legend().remove() if ax2.get_legend() else None

# Title and layout
plt.title("Top 10 Pages - High Session and Engaged Sessions per Active User", fontsize=14)
plt.tight_layout()
plt.show()








```
